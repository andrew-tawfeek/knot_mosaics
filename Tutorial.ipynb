```python
# To generate a random suitably connected knot mosaic of dimension 5:
M = random_mosaic(dimension = 5)

# To get its matrix representation:
M.matrix()
```

```python
# To see what it looks like:
M.show(4) # the 4 is the size of each tile in the mosaic
```

```python
# You have some optional inputs for random mosaics as well:
M = random_mosaic(dimension = 5, suitably_connected=True, num_crossings = 3)
M.show(4)
```

```python
# Given a mosaic M, you can find its crossings:
M.findCrossings()
```

```python
# Then beginning at a crossing, you can "walk" along the knot:
M.strandOf((2,1), direction = 'right')

# this will keep walking until the tile right before the beginning tile
```

```python
# You can also specify a mosaic with a matrix, tile by tile:
hopf = Mosaic([[0,2,1,0],[2,9,10,1],[3,10,10,4],[0,3,4,0]]); hopf.show(3)
```

```python
hopf.isSuitablyConnected() # checking that it is connected
```

```python
hopf.directions(0,1) # what directions you can go from the (0,1) tile -- mosaics are 0-indexed
```

```python
# You can also generate random mosaics that are not suitably connected:
random_mosaic(dimension = 5, suitably_connected=False).show(4)
```

```python
# Lastly, you can "zoom" into a mosaic, expanding each tile to a 3x3:
hopf.zoom().show(8)
```
