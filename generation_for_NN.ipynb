```python
# generating data.py -- the training data

samples = 10000

data_pairs = [(list(random_mosaic(dimension = 5).matrix()) ,1) for i in range(0,samples)] + [(list(random_mosaic(dimension = 5, suitably_connected=False).matrix()) ,0) for i in range(0,samples)]

with open('data.py', 'w') as f:
    f.write('data_pairs = ' + str(data_pairs))
```

```python
# generating unseen_data.py -- the unseen data, i.e. not in training set
# every mosaic generated is suitably connected!

samples = 1000

unseen_data_pairs = [(list(random_mosaic(dimension = 5, suitably_connected=True).matrix()) ,1) for i in range(0,samples)]

with open('unseen_data.py', 'w') as f:
    f.write('unseen_data_pairs = ' + str(unseen_data_pairs))
```
