```python
# generating data.py -- the training data
# pairs of matrix representation of mosaics and 0 or 1 depending on whether the mosaic is suitably connected or not

samples = 100
dimension = 10

data_pairs = [(list(random_mosaic(dimension = dimension).matrix()),1) for i in range(0,samples)] + [(list(random_mosaic(dimension = dimension, suitably_connected=False).matrix()) ,0) for i in range(0,samples)]

with open('data.py', 'w') as f:
    f.write('data_pairs = ' + str(data_pairs))
```

```python
# generating unseen_data.py -- the unseen data, i.e. not in training set

samples = 500
dimension = 10

unseen_data_pairs = [(list(random_mosaic(dimension = dimension, suitably_connected=True).matrix()) ,1) for i in range(0,samples)] + [(list(random_mosaic(dimension = dimension, suitably_connected=False).matrix()) ,0) for i in range(0,samples)]
shuffle(unseen_data_pairs)

with open('unseen_data.py', 'w') as f:
    f.write('unseen_data_pairs = ' + str(unseen_data_pairs))
```
