```sage
def check_unknot(knot_pd):
    try:
        L = snappy.Link(knot_pd) # Using snappy and regina to check topology
        M = L.exterior()
        T = regina.Triangulation3(M)
        if T.isSolidTorus():
            return true
        else:
            return false
    except ValueError:
        print("ValueError raised")
        print("knot mosaic: " + knot_mosaic)
        print("oriented gauss code: " + orientedGaussCode(knot_mosaic))
        check_unknot(knot_mosaic)    
```

```sage
def generate_unknotJones(dimension,sampleSize,fileName):
    # 0 indicates trivial, 1 indicates non-trivial
    data = []
    for i in range(sampleSize):
        unknot = -1
        jones = []

        M = random_mosaic(dimension = dimension, num_components=1)
        sageMatrix = M.matrix()
        sageLink = Link(orientedGaussCode(M)) # 

        # converting the sageMath matrix of sagemath integers to a nested list of python integers
        json_matrix = [[int(sageMatrix[i, j]) for j in range(sageMatrix.ncols())] for i in range(sageMatrix.nrows())] 

        # Checking conditions for unknot
        if(M.numCrossings() < 3):
            unknot = '0'
        elif(check_unknot(sageLink.pd_code())):
            unknot = '0'
        else:
            unknot = '1'

        # Compute Jones polynomial
        jones = Matrix(sageLink.jones_polynomial(algorithm = 'statesum').coefficients())
        json_jones = [[int(jones[i, j]) for j in range(jones.ncols())] for i in range(jones.nrows())] 

        data.append({'matrix': json_matrix, 'unknot': unknot, 'jones_coeff': json_jones})

    with open(fileName, 'w') as f:
        json.dump(data, f, indent=2)
```
